```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# y = mx + b || m is slope, b is y-intercept, x is independent variable, y is dependent variable 
np.random.seed(42)
```

```python
num_samples = 500
yrs_exp = np.random.randint(2, 21, size=num_samples)
yrs_exp
```

```python
slope = (200000 + 60000) / 18
slope
```

```python
np.random.normal(0, 10000, size=num_samples)
```

```python
intercept = 60000
salaries = slope * yrs_exp + intercept + np.random.normal(0, 10000, size=num_samples)
salaries
```

```python
df = pd.DataFrame({'YearsExperience': yrs_exp, 'Salary': salaries})
df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='YearsExperience', y='Salary', data=df, color='blue')
sns.regplot(x='YearsExperience', y='Salary', data=df, scatter=False, color='red')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Linear Regression:Salary vs Years of Experience')
plt.show()
```

```python
X = df[['YearsExperience']]
y = df['Salary']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13)
```

```python
model_LR = LinearRegression()
```

```python
model_LR.fit(X_train, y_train)
```

```python
model_LR.score(X_train, y_train)
```

```python
model_LR.score(X_test, y_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
y_pred = model_LR.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

```python
# y = mx + b || m is slope: coeffiecient of x, b is y-intercept, x is independent variable, y is dependent variable 
model_LR.coef_
```

```python
model_LR.intercept_
```

```python
coeffs = model_LR.coef_
intercept = model_LR.intercept_

X = np.linspace(0, 20, 100)
y = coeffs * X + intercept

plt.scatter(X, y, color='blue', label=f'y = {coeffs[0]}x + {intercept}')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Linear Regression Salary')
plt.grid()
plt.show() 
```
