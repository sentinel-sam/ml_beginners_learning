Linear regression is a way to predict the value of some a target variable by fitting a line that best describes the relationship between Big X and little y for the values we already have.
y = mx + b

When we train a model, we're solving an optimization problem. We provide training data to an algorithm and tell it to find the model or model parameters that best fit the data.
Cost Function(loss or error function): a mathematical formula that provides the score by which the algorithm will determine the best fit.

Generally, the the goal is to minimize the cost function and get the lowest score. For linear models, these functions measure distance, and the model tries to to get the closest fit to the data. For tree-based models, they measure impurity, and the model tries to get the most terminal nodes.

Residuals: the vertical distance between each individual data point(the actual data that came from real observations) and the regression line. Each data point has only one residual which can be positive if it’s above the regression line, negative if it’s below the regression line, or zero if the line passes directly through the point
The model describes theoretical line which doesn't really exist outside the model.


Error = Reality - Measurement

mean absolute error (MAE). Absolute error is the error in a single measurement, and mean absolute error is the average error over the course of several measurements.

Leakage is the use of data in training your model that would not be typically be available when making predictions

Imputation = filling missing values by extrapolation using the data patterns, or mean values, or ignore missing values if there are few missing values in a large dataset.

Generalization => testing model with different data than you used to train it
